# Session 2 — Homework Solution: Password Cracker

Brute-force solution to last session's HW.

```python
ascii_chars = [chr(i) for i in range(32, 127)]
target_hash = 2868533088

def hash_password(text):
    hash = 0
    for char in text:
        hash = (hash * 281 ^ ord(char) * 997) & 0xFFFFFFFF
    return hash
```

For each candidate password, we apply the same hash_password() function (already provided). \
So we need to compare all candidate with the target hash, 2868533088. \
To make the search slightly more efficient, we reuse the intermediate hash after each character instead of recomputing the full password hash from scratch every time.

```python
def crack_password(target):
    # Brute force all 4-character combinations.
    # We reuse the intermediate hash to avoid recomputing the prefix each time.
    for a in ascii_chars:
        h1 = (ord(a) * 997) & 0xFFFFFFFF
        for b in ascii_chars:
            h2 = (h1 * 281 ^ ord(b) * 997) & 0xFFFFFFFF
            for c in ascii_chars:
                h3 = (h2 * 281 ^ ord(c) * 997) & 0xFFFFFFFF
                for d in ascii_chars:
                    h4 = (h3 * 281 ^ ord(d) * 997) & 0xFFFFFFFF
                    if h4 == target:
                        return a + b + c + d

password = crack_password(target_hash)
print(f"Password found: {password!r}")
```

```python
assert hash_password(password) == target_hash
print("Hash verified.")
```

**Small note.** The supplied `ascii_chars` list contains the 95 printable ASCII characters (`32` through `126`). The search space is `95**4 = 81,450,625`.
