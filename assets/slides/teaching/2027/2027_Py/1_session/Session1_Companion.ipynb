## 3. Errors, exceptions, and warnings
### 3.1 Deliberately producing an error (check slide "Errors")

Errors are part of programming. Here we intentionally divide by zero and explicitly raise a `ValueError`, but catch both so the lesson can continue.

```python
position_value = 250
number_of_shares = 0

try:
    average_price = position_value / number_of_shares
except ZeroDivisionError as error:
    print(type(error).__name__ + ":", error)

try:
    raise ValueError("A demo value was declared invalid on purpose.")
except ValueError as error:
    print(type(error).__name__ + ":", error)
```

### 6.2 Common list methods (check slide "Most common list methods")

List methods can modify a list in place. We work on a copy so the original price history remains available for later examples.

```python
working_prices = prices.copy()
print("first element:", working_prices[0])

working_prices.append(52.00)
working_prices.extend([52.40, 52.10])
working_prices.insert(0, 48.00)
print("after additions:", working_prices)

working_prices.remove(48.00)
last_observation = working_prices.pop()
first_observation = working_prices.pop(0)
print("after removals:", working_prices)
print("popped values:", last_observation, first_observation)

working_prices.clear()
print("after clear:", working_prices)
```

### 7.3 `for` loops (check slide "For loops")

### 7.4 `range`, `enumerate`, and list comprehensions (check slide "For loops, range, and enumerate")

`range` generates indices, `enumerate` gives both an index and a value, and a list comprehension creates a list compactly.

```python
for index in range(len(prices)):
    print("index:", index)

for index, price in enumerate(prices):
    print(index, price)

price_changes = [price - prices[0] for price in prices]
print("changes from first price:", price_changes)
```

### 7.5 `zip` (check slide "zip")

```python
days = ["Mon", "Tue", "Wed", "Thu"]

for day, price in zip(days, prices):
    print(day, price)

for index, (day, price) in enumerate(zip(days, prices)):
    print(index, day, price)
```

### 7.7 `break` and `continue` (check slide "Break/Continue statements")

`break` exits a loop early; `continue` skips the rest of the current iteration. Both let us control a loop without changing its basic structure.

```python
for price in [48.5, 49.1, 50.2, 51.25, 52.0]:
    if price >= 51:
        print("First price at or above 51:", price)
        break
    print("still below 51:", price)
```

### 9.4 `*args` and `**kwargs` (check slide "args and kwargs")

`*args` collects positional arguments, while `**kwargs` collects named arguments. This is useful when the number of inputs is not fixed in advance.

```python
def total_cost(*costs):
    return sum(costs)

print(total_cost(2.50, 1.25, 0.75))


def total_positions(**positions):
    return sum(positions.values())

print(total_positions(NOVA=615.0, LUMA=220.0, PINE=180.0))


def total_everything(*costs, **positions):
    return sum(costs) + sum(positions.values())

print(total_everything(2.50, 1.25, NOVA=615.0, LUMA=220.0))
```

## 14. Decorators
### 14.1 A first decorator (check slide "Decorators")

A decorator receives a function and returns a modified function. This simple decorator prints before and after the original function runs.

```python
def announce(func):
    def wrapper(*args, **kwargs):
        print("Starting calculation...")
        result = func(*args, **kwargs)
        print("Calculation finished.")
        return result
    return wrapper


@announce
def value_position(price, shares):
    return price * shares

print("result:", value_position(51.25, 12))
```

### 14.2 Timing a function (check slide "Timing a function")

A timing decorator measures how long a function takes. `timeit` is useful when we want more systematic timing of a small expression.

```python
import time
import timeit


def timer(func):
    def wrapper(*args, **kwargs):
        start = time.time()
        result = func(*args, **kwargs)
        end = time.time()
        print(f"{func.__name__} took {end - start:.6f} seconds")
        return result
    return wrapper


@timer
def compound_once(value, rate):
    return value * (1 + rate)

print(compound_once(100, 0.05))
print("timeit result:", timeit.timeit("100 * 1.05", number=10_000))
```

### 14.3 Logging a function with a decorator (check slide "Useful decorators: Logging a function")

The decorator can also record which function is running and with which arguments. We keep the underlying valuation function unchanged.

```python
def log_call(func):
    def wrapper(*args, **kwargs):
        logging.info(
            "Running %s with args=%s kwargs=%s",
            func.__name__,
            args,
            kwargs,
        )
        return func(*args, **kwargs)
    return wrapper


@log_call
def net_value(price, shares, fee=0.0):
    return price * shares - fee

print(net_value(51.25, 12, fee=2.50))
```

## 17. Floating-point precision
### 17.1 Machine epsilon (check slide "Machine Epsilon")

Machine epsilon describes the spacing of floating-point numbers around 1.0. It helps explain why adding an extremely small quantity can sometimes leave a float unchanged.

```python
epsilon = sys.float_info.epsilon
print("machine epsilon:", epsilon)
print(1.0 + epsilon == 1.0)
print(1.0 + epsilon / 2 == 1.0)

# A famous finite-precision example:
print("0.1 + 0.2 =", 0.1 + 0.2)
print("exactly 0.3?", 0.1 + 0.2 == 0.3)
```
